## 1. Classical Probability

Recall:

\[
P(A)=\frac{\text{favourable outcomes}}{\text{total outcomes}}
\]

Example: What is the probability of rolling an even number on a fair die?

```python
# Working Example: Classical Probability

sample_space = list(range(1, 7))

# Event: rolling an even number
event_even = [x for x in sample_space if x % 2 == 0]

probability_even = len(event_even) / len(sample_space)

print("Sample space:", sample_space)
print("Even outcomes:", event_even)
print("P(Even) =", probability_even)
```

## 2. Complement Rule

The lecture gives:

\[
P(A') = 1-P(A)
\]

This is especially useful for **"at least one"** problems.

Example: Find the probability of getting at least one head in 5 fair coin flips.

Instead of calculating all possibilities containing a head, calculate:

\[
P(\text{at least one head})=1-P(\text{no heads})
\]



```python
# Working Example: At least one head in 5 flips

p_head = 0.5
number_of_flips = 5

p_no_heads = (1 - p_head) ** number_of_flips
p_at_least_one_head = 1 - p_no_heads

print("P(no heads) =", p_no_heads)
print("P(at least one head) =", p_at_least_one_head)
```

## 3. Addition Rule — A OR B

The addition rule is:

\[
P(A\cup B)=P(A)+P(B)-P(A\cap B)
\]

We subtract the intersection because it is counted twice.

Example: Drawing one card from a standard 52-card deck.

- A = card is a Heart → 13 cards
- B = card is a King → 4 cards
- A AND B = King of Hearts → 1 card



```python
# Working Example: Heart OR King

total_cards = 52
hearts = 13
kings = 4
heart_and_king = 1

p_heart = hearts / total_cards
p_king = kings / total_cards
p_heart_and_king = heart_and_king / total_cards

p_heart_or_king = p_heart + p_king - p_heart_and_king

print("P(Heart) =", p_heart)
print("P(King) =", p_king)
print("P(Heart AND King) =", p_heart_and_king)
print("P(Heart OR King) =", p_heart_or_king)
```

## 4. Multiplication Rule and Independence

General multiplication rule:

\[
P(A\cap B)=P(A)P(B|A)
\]

For independent events:

\[
P(A\cap B)=P(A)P(B)
\]

Example: Two fair coin flips.

The result of the first flip does not affect the second flip.



```python
# Working Example: Two heads in two independent flips

p_head_first = 0.5
p_head_second_given_first = 0.5

p_two_heads = p_head_first * p_head_second_given_first

print("P(H1 AND H2) =", p_two_heads)
```

## 5. Mutually Exclusive Events

Two events are mutually exclusive if they **cannot happen at the same time**.

For mutually exclusive events:

\[
P(A\cap B)=0
\]

Example:

- A = rolling a 2
- B = rolling a 5

On one die roll, both cannot happen.

**Think:** Independence asks *"Does one event affect the other?"*  
Mutual exclusivity asks *"Can both happen?"*

```python
# Working Example: Mutually exclusive events

sample_space = set(range(1, 7))

A = {2}
B = {5}

intersection = A.intersection(B)

print("A AND B =", intersection)
print("Mutually exclusive?", len(intersection) == 0)
```

## 6. Permutations vs. Combinations

The key question is:

> **Does order matter?**

### Permutation

Order matters:

\[
P(n,r)=\frac{n!}{(n-r)!}
\]

Example: 10 runners, choose gold, silver, and bronze.

### Combination

Order does not matter:

\[
C(n,r)=\frac{n!}{r!(n-r)!}
\]

Example: Choose 4 students from 10 for a committee.


```python
# Working Example: Permutation and Combination

import math

def permutation(n, r):
    return math.factorial(n) // math.factorial(n - r)

def combination(n, r):
    return math.factorial(n) // (
        math.factorial(r) * math.factorial(n - r)
    )

print("P(10, 3) =", permutation(10, 3))
print("C(10, 4) =", combination(10, 4))
```

## 7. Joint, Marginal, and Conditional Probability

Consider this survey of 1,000 people:

|              | Regular Exercise | Irregular Exercise | Total |
|--------------|-----------------:|--------------------:|------:|
| Male         | 0.30             | 0.20                | 0.50  |
| Female       | 0.35             | 0.15                | 0.50  |
| Total        | 0.65             | 0.35                | 1.00  |

### Joint

An inner cell:

\[
P(\text{Female AND Regular})=0.35
\]

### Marginal

A row/column total:

\[
P(\text{Female})=0.50
\]

### Conditional

Narrow the sample space:

\[
P(\text{Regular}|\text{Female})
=
\frac{P(\text{Regular AND Female})}
{P(\text{Female})}
\]



```python
# Working Example: Conditional Probability

p_female_and_regular = 0.35
p_female = 0.50

p_regular_given_female = (
    p_female_and_regular / p_female
)

print("P(Regular | Female) =", p_regular_given_female)
```

## 8. Bayes' Theorem

Bayes' Theorem:

\[
P(H|E)=\frac{P(E|H)P(H)}{P(E)}
\]

The four important ideas:

- **Prior:** P(H)
- **Likelihood:** P(E|H)
- **Marginal evidence:** P(E)
- **Posterior:** P(H|E)

Bayes lets us **update our belief after seeing evidence**.

```python
# Working Example: Spam Filter

def bayes(prior, likelihood, marginal):
    return (likelihood * prior) / marginal

p_spam = 0.30
p_free_given_spam = 0.80
p_free_given_legit = 0.05

# Total probability of seeing "FREE"
p_free = (
    p_free_given_spam * p_spam
    + p_free_given_legit * (1 - p_spam)
)

p_spam_given_free = bayes(
    p_spam,
    p_free_given_spam,
    p_free
)

print("P(FREE) =", p_free)
print("P(Spam | FREE) =", p_spam_given_free)
```

### Your Turn — Bayes and Manufacturing

A factory produces:

- 60% of parts from Machine A
- 40% from Machine B

Defect rates:

- P(Defective | A) = 0.02
- P(Defective | B) = 0.05

A randomly selected part is found to be defective.

Calculate:

\[
P(B|Defective)
\]

Your steps should be:

1. Calculate P(Defective).
2. Apply Bayes' Theorem.
3. Interpret your answer in one sentence.

Use the `bayes()` function.

```python
def bayes(prior, likelihood, marginal):
    return (likelihood * prior) / marginal

p_A = 0.60
p_B = 0.40

p_defective_given_A = 0.02
p_defective_given_B = 0.05

p_defective = (p_defective_given_A * p_A) + (p_defective_given_B * p_B)

p_B_given_defective = bayes(p_B, p_defective_given_B, p_defective)

print("P(Defective) =", p_defective)
print("P(Machine B | Defective) =", p_B_given_defective)
```
