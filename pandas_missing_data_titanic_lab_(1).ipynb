## Section 7 — Handle Missing Data: Imputation
⏱ ~10 min

Imputation fills missing values with an estimated substitute instead of removing data. Work on `df3 = df.copy()` throughout this section.

**Quick reference — choosing the right strategy:**

| Data type | Strategy |
|---|---|
| Numeric, symmetric | mean |
| Numeric, skewed / outliers | median |
| Categorical / text | mode (most frequent) |
| Time-series / ordered | ffill or bfill |
| Truly unknown | `'Unknown'` / `'N/A'` / `0` |

### Exercise 3 — Impute Missing Values  ⏱ 10 min

> Start with `df3 = df.copy()` and work on `df3` throughout.
> **Hint:** Use `df['col'].mode()[0]` to get the most common category value.

**Tasks:**
1. Fill all `Math` NaNs with the **mean** of the Math column. Print the updated Math column.
   - Expected mean: `≈ 79.57`
2. Fill all `Science` NaNs with the **median** of the Science column.
   - Expected median: `80.0`
3. Fill all `English` NaNs with **forward fill** (`ffill`). Then check — are any English NaNs left?
   - If the first row is NaN, `ffill` cannot fill it. Chain with `bfill` to catch it.
4. Fill all `City` NaNs with the **most frequent city** (mode).
5. Fill all `Grade` NaNs with the string `'N/A'`.
6. Fill all `Passed` NaNs with `False`. Confirm with `isnull().sum()` on the `Passed` column.

> **Why does ffill sometimes leave NaN?**
> `ffill` copies the *previous* row's value. If the first row of the column is NaN, there is no previous value — so it stays NaN.
> Fix: `df['col'] = df['col'].ffill().bfill()`

````python
df3 = df.copy()

# 1. Math mean
df3['Math'] = df3['Math'].fillna(df3['Math'].mean())
print("Updated Math column:")
print(df3['Math'])

# 2. Science median
df3['Science'] = df3['Science'].fillna(df3['Science'].median())

# 3. English ffill + bfill
df3['English'] = df3['English'].ffill().bfill()
print("\nEnglish NaNs left:", df3['English'].isnull().sum())

# 4. City mode
df3['City'] = df3['City'].fillna(df3['City'].mode()[0])

# 5. Grade 'N/A'
df3['Grade'] = df3['Grade'].fillna('N/A')

# 6. Passed False
df3['Passed'] = df3['Passed'].fillna(False)
print("\nPassed column NaNs remaining:", df3['Passed'].isnull().sum())
````

## Section 8 — Full Cleaning Pipeline
⏱ ~8 min

Put everything together. Start fresh with `df4 = df.copy()` and clean the entire dataset in a single pipeline. **No NaN should remain when you are done.**

### Exercise 4 — Full Cleaning Pipeline  ⏱ 8 min 🏆

> **Hint:** Use a `for` loop for Step 1 to fill multiple numeric columns neatly.

**Tasks:**
1. Fill `Math` NaNs with the column **mean**. Fill `Science` NaNs with the column **median**.
2. Fill `English` NaNs with **forward fill then backward fill** (to handle edge rows).
3. Fill `City` NaNs with the most frequent city (**mode**).
4. Fill `Grade` NaNs with `'N/A'` and `Passed` NaNs with `False`.
5. Add a new column `Total = Math + Science + English` (round to 1 decimal).
6. Assert that **zero NaN values remain** in the entire DataFrame, then print a final summary table.

**Expected final output (first 3 rows):**
```
Dataset is fully clean. No missing values.
   Name  Math  Science  English      City Grade  Passed  Total
0   Ali  85.0     78.0     88.0    Lahore     A    True  251.0
```

### Q12 — DateTime Operations  ⏱ 6 min

The Titanic sank on **April 15, 1912**. Let's engineer date-based features from that fixed event.

**Tasks:**
1. Create column `EventDate` with the value `'1912-04-15'` for every row. Convert it to datetime with `pd.to_datetime()`.
2. Create column `YearsAgo` = the number of **complete years** between today and `EventDate`. Use `pd.Timestamp.now()` and `.dt.days`.

````python
titanic['EventDate'] = pd.to_datetime('1912-04-15')

today = pd.Timestamp.now()
titanic['YearsAgo'] = (today - titanic['EventDate']).dt.days // 365

titanic['EventMonth'] = titanic['EventDate'].dt.month
titanic['EventDayOfWeek'] = titanic['EventDate'].dt.dayofweek

sink_day = pd.Timestamp('1912-04-15').day_name()
print(f"The Titanic sank on a {sink_day}.")
````
