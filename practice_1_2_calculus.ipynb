# Exercise 1: Derivatives — Basic Rules

A **derivative** tells you the slope of a function at any point — how fast is it changing?

We can approximate it numerically using the **central difference formula**:

```
f'(x)  ≈  (f(x + h) - f(x - h)) / (2 * h)

where h is a very small number (e.g. 0.00001)
```

---

## Common Derivative Rules

```
f(x) = c          =>  f'(x) = 0            (constant)
f(x) = x^n        =>  f'(x) = n * x^(n-1)  (power rule)
f(x) = e^x        =>  f'(x) = e^x
f(x) = ln(x)      =>  f'(x) = 1/x
f(x) = f + g      =>  f'(x) = f' + g'      (sum rule)
f(x) = f * g      =>  f'(x) = f'g + fg'    (product rule)
```

---

## Examples

```
f(x) = x^3        =>  f'(x) = 3x^2
f(x) = 4x^2 + 2x  =>  f'(x) = 8x + 2
f(x) = x^2 * e^x  =>  f'(x) = 2x*e^x + x^2*e^x   (product rule)
```

````python

import numpy as np


def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)


def f1(x):
    return x**3


def f1_prime(x):
    return 3 * x**2


def f2(x):
    return 4 * x**2 + 2 * x


def f2_prime(x):
    return 8 * x + 2


def f3(x):
    return np.exp(x)


def f3_prime(x):
    return np.exp(x)


def f4(x):
    return np.log(x)


def f4_prime(x):
    return 1 / x


x = 2.0
print(f"At x = {x}")
print(
    f"  f1 = x^3         analytic={f1_prime(x):.4f}  numerical={numerical_derivative(f1, x):.4f}"
)
print(
    f"  f2 = 4x^2+2x     analytic={f2_prime(x):.4f}  numerical={numerical_derivative(f2, x):.4f}"
)
print(
    f"  f3 = e^x         analytic={f3_prime(x):.4f}  numerical={numerical_derivative(f3, x):.4f}"
)
print(
    f"  f4 = ln(x)       analytic={f4_prime(x):.4f}  numerical={numerical_derivative(f4, x):.4f}"
)
````

# Exercise 2: Plot a Function and Its Derivative

The derivative tells us the **slope** at every point:

```
f'(x) > 0   =>  function is going UP
f'(x) < 0   =>  function is going DOWN
f'(x) = 0   =>  function is at a peak or a valley  (minimum or maximum)
```

---

## Example with f(x) = x³ − 3x

```
f'(x) = 3x² - 3

Setting f'(x) = 0:
  3x² - 3 = 0
  x² = 1
  x = 1  or  x = -1     <-- these are the critical points
```

At x = -1: local maximum  
At x =  1: local minimum

````python

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-2.5, 2.5, 300)


def f_ex2(x):
    return x**3 - 3 * x


def f_prime_ex2(x):
    return 3 * x**2 - 3


critical_x = [-1.0, 1.0]

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(7, 6), sharex=True)

ax1.plot(x, f_ex2(x), color="steelblue", lw=2, label="f(x) = x³ - 3x")
for cx in critical_x:
    ax1.scatter([cx], [f_ex2(cx)], color="red", zorder=5, s=80)
    ax1.axvline(cx, color="red", lw=0.8, linestyle=":")
ax1.axhline(0, color="grey", lw=0.5)
ax1.set_ylabel("f(x)")
ax1.legend()
ax1.grid(True, alpha=0.3)
ax1.set_title("Function and its Critical Points")

ax2.plot(x, f_prime_ex2(x), color="darkorange", lw=2, label="f'(x) = 3x² - 3")
ax2.axhline(0, color="red", lw=1, linestyle="--", label="f'(x) = 0")
for cx in critical_x:
    ax2.scatter([cx], [0], color="red", zorder=5, s=80)
ax2.set_xlabel("x")
ax2.set_ylabel("f'(x)")
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

# Exercise 3: Activation Function Derivatives

These three functions are used inside neural networks. Their derivatives are used during training.

---

## 1. 🔵 Sigmoid

```
σ(x) = 1 / (1 + e^(-x))

σ'(x) = σ(x) * (1 - σ(x))
```

Output is always between 0 and 1. Derivative is highest at x = 0.

---

## 2. 🟠 ReLU (Rectified Linear Unit)

```
ReLU(x) = max(0, x)

ReLU'(x) = 0  if x < 0
           1  if x > 0
```

Very simple: just cuts off negative values. Most popular activation in modern networks.

---

## 3. 🟢 Tanh

```
tanh(x) = (e^x - e^(-x)) / (e^x + e^(-x))

tanh'(x) = 1 - tanh(x)^2
```

Output is between -1 and 1. Stronger gradient than sigmoid.

````python
# Exercise 3
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 400)


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def sigmoid_prime(x):
    s = sigmoid(x)
    return s * (1 - s)


def relu(x):
    return np.maximum(0, x)


def relu_prime(x):
    return np.where(x > 0, 1.0, 0.0)


def tanh(x):
    return (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))


def tanh_prime(x):
    return 1 - tanh(x) ** 2


fig, axes = plt.subplots(1, 3, figsize=(14, 4))
configs = [
    (sigmoid, sigmoid_prime, "Sigmoid", "steelblue"),
    (relu, relu_prime, "ReLU", "darkorange"),
    (tanh, tanh_prime, "Tanh", "green"),
]

for ax, (fn, fn_p, name, color) in zip(axes, configs):
    ax.plot(x, fn(x), color=color, lw=2, label=name)
    ax.plot(
        x,
        fn_p(x),
        color=color,
        lw=2,
        label=f"{name}'",
        linestyle="--",
        alpha=0.7,
    )
    ax.axhline(0, color="grey", lw=0.5)
    ax.axvline(0, color="grey", lw=0.5)
    ax.set_title(name)
    ax.legend(fontsize=8)
    ax.grid(True, alpha=0.3)
    ax.set_ylim(-1.5, 1.5)

plt.suptitle("Activation Functions and Their Derivatives", fontweight="bold")
plt.tight_layout()
plt.show()
````

# Exercise 4: The Chain Rule

When one function is **inside** another, use the chain rule to find the derivative.

```
If  h(x) = f(g(x))  then:

  h'(x) = f'(g(x)) × g'(x)

  "derivative of the outside × derivative of the inside"
```

---

## Example 1 — Two functions

```
h(x) = (3x² + 1)⁵

  inner:  g(x) = 3x² + 1     =>  g'(x) = 6x
  outer:  f(u) = u⁵           =>  f'(u) = 5u⁴

  h'(x) = 5(3x²+1)⁴ × 6x  =  30x(3x²+1)⁴
```

---

## Example 2 — Three functions (common in neural networks)

```
h(x) = ln(sigmoid(3x))

  innermost:  k(x) = 3x            =>  k'(x) = 3
  middle:     g(v) = sigmoid(v)     =>  g'(v) = sigmoid(v)*(1-sigmoid(v))
  outer:      f(u) = ln(u)          =>  f'(u) = 1/u

  h'(x) = f'(g(k(x))) × g'(k(x)) × k'(x)
         =  (1/sigmoid(3x))  ×  sigmoid(3x)*(1-sigmoid(3x))  ×  3
         =  3 * (1 - sigmoid(3x))
```

````python
# Exercise 4
import numpy as np


def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def h1(x):
    return sigmoid(x**2 + 1)


def h1_prime(x):
    u = x**2 + 1
    return sigmoid(u) * (1 - sigmoid(u)) * (2 * x)


def h2(x):
    return np.log(sigmoid(3 * x))


def h2_prime(x):
    return 3 * (1 - sigmoid(3 * x))


test_points = [-2.0, -1.0, 0.0, 1.0, 2.0]
print("h1(x) = sigmoid(x^2 + 1)")
for x_val in test_points:
    a = h1_prime(x_val)
    n = numerical_derivative(h1, x_val)
    print(
        f"  x={x_val:5.1f}   analytic={a:.5f}   numerical={n:.5f}   match={np.isclose(a, n)}"
    )

print("\nh2(x) = ln(sigmoid(3x))")
for x_val in test_points:
    a = h2_prime(x_val)
    n = numerical_derivative(h2, x_val)
    print(
        f"  x={x_val:5.1f}   analytic={a:.5f}   numerical={n:.5f}   match={np.isclose(a, n)}"
    )
````

# Exercise 5: Partial Derivatives

When a function has **more than one input**, a partial derivative measures the effect of changing just **one** variable while keeping all others fixed.

```
∂f/∂x  means: differentiate f with respect to x, treat all other variables as constants.
```

---

## Example

```
f(x, y) = x² + 3xy + y²

∂f/∂x = 2x + 3y     (treat y as a constant number)
∂f/∂y = 3x + 2y     (treat x as a constant number)
```

At the point (x=2, y=3):
```
∂f/∂x = 2(2) + 3(3) = 4 + 9  = 13
∂f/∂y = 3(2) + 2(3) = 6 + 6  = 12
```

---

## Numerical partial derivative

We can approximate it the same way as before — nudge one variable by h, keep others fixed:
```
∂f/∂x  ≈  (f(x+h, y) - f(x-h, y)) / (2h)
∂f/∂y  ≈  (f(x, y+h) - f(x, y-h)) / (2h)
```

````python
# Exercise 5
import numpy as np


def f1(x, y):
    return x**2 + 3 * x * y + y**2


def df1_dx(x, y):
    return 2 * x + 3 * y


def df1_dy(x, y):
    return 3 * x + 2 * y


def f2(x, y):
    return (x**3) * y + 2 * x * (y**2)


def df2_dx(x, y):
    return 3 * (x**2) * y + 2 * (y**2)


def df2_dy(x, y):
    return x**3 + 4 * x * y


def f3(x, y):
    return np.exp(x * y)


def df3_dx(x, y):
    return y * np.exp(x * y)


def df3_dy(x, y):
    return x * np.exp(x * y)


h = 1e-5
x, y = 2.0, 3.0
print(f"At (x={x}, y={y})\n")
print("f1 = x^2 + 3xy + y^2")
print(
    f"  df/dx  analytic={df1_dx(x,y):.4f}  numerical={(f1(x+h,y)-f1(x-h,y))/(2*h):.4f}"
)
print(
    f"  df/dy  analytic={df1_dy(x,y):.4f}  numerical={(f1(x,y+h)-f1(x,y-h))/(2*h):.4f}"
)

print("\nf2 = x^3*y + 2*x*y^2")
print(
    f"  df/dx  analytic={df2_dx(x,y):.4f}  numerical={(f2(x+h,y)-f2(x-h,y))/(2*h):.4f}"
)
print(
    f"  df/dy  analytic={df2_dy(x,y):.4f}  numerical={(f2(x,y+h)-f2(x,y-h))/(2*h):.4f}"
)

print("\nf3 = exp(x*y)")
print(
    f"  df/dx  analytic={df3_dx(x,y):.4f}  numerical={(f3(x+h,y)-f3(x-h,y))/(2*h):.4f}"
)
print(
    f"  df/dy  analytic={df3_dy(x,y):.4f}  numerical={(f3(x,y+h)-f3(x,y-h))/(2*h):.4f}"
)
````

# Exercise 6: The Gradient Vector

The **gradient** is just a vector that collects ALL the partial derivatives together:

```
∇f(x, y) = [∂f/∂x,  ∂f/∂y]

∇f(x, y, z) = [∂f/∂x,  ∂f/∂y,  ∂f/∂z]
```

---

## What does it mean?

```
∇f   points in the direction of STEEPEST INCREASE
-∇f  points in the direction of STEEPEST DECREASE  ← this is what we use in training

At a minimum:  ∇f = [0, 0, ...]  (zero vector)
```

---

## Example

```
f(x, y) = x² + 2y²

∂f/∂x = 2x
∂f/∂y = 4y

∇f(x, y) = [2x, 4y]

At (3, 1):  ∇f = [6, 4]   -- gradient points away from minimum at (0,0)
```

---

## Gradient Descent Update Rule

```
w_new = w_old - α × ∇f(w_old)

α  = learning rate (small positive number, e.g. 0.1)
```

Subtract the gradient → move downhill → reduce the function value.

````python
# Exercise 6
import matplotlib.pyplot as plt
import numpy as np


def f(x, y):
    return x**2 + 2 * y**2


def gradient_f_ex6(x, y):
    df_dx = 2 * x
    df_dy = 4 * y
    return np.array([df_dx, df_dy])


def gradient_descent(x_start, y_start, alpha=0.1, steps=30):
    w = np.array([x_start, y_start])
    path = [w.copy()]
    values = [f(*w)]
    for _ in range(steps):
        g = gradient_f_ex6(*w)
        w = w - alpha * g
        path.append(w.copy())
        values.append(f(*w))
    return np.array(path), values


path, values = gradient_descent(2.5, 2.0, alpha=0.1, steps=30)
print(f"Start: x={path[0,0]:.2f}, y={path[0,1]:.2f}, f={values[0]:.4f}")
print(f"End:   x={path[-1,0]:.4f}, y={path[-1,1]:.4f}, f={values[-1]:.6f}")
print("True minimum is at (0, 0) where f = 0")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

xr = np.linspace(-3, 3, 200)
yr = np.linspace(-3, 3, 200)
XX, YY = np.meshgrid(xr, yr)
ZZ = XX**2 + 2 * YY**2
ax1.contour(XX, YY, ZZ, levels=15, cmap="Blues", alpha=0.7)
ax1.plot(
    path[:, 0],
    path[:, 1],
    "o-",
    color="darkorange",
    lw=2,
    ms=4,
    label="Descent path",
)
ax1.scatter(*path[0], color="green", s=100, zorder=5, label="Start")
ax1.scatter(*path[-1], color="red", s=100, zorder=5, label="End")
ax1.set_xlabel("x")
ax1.set_ylabel("y")
ax1.set_aspect("equal")
ax1.grid(True, alpha=0.3)
ax1.legend()
ax1.set_title("Path on Contour Plot")

ax2.plot(values, color="steelblue", lw=2, marker="o", ms=4)
ax2.set_xlabel("Step")
ax2.set_ylabel("f(x, y)")
ax2.set_title("Function Value Per Step")
ax2.grid(True, alpha=0.3)

plt.suptitle("Gradient Descent on f(x,y) = x² + 2y²", fontweight="bold")
plt.tight_layout()
plt.show()
````

# Exercise 7: Effect of Learning Rate

The learning rate **α** controls how big each step is.

```
Too small  →  takes many steps, very slow to converge
Too large  →  overshoots the minimum, may never converge
Just right →  reaches the minimum smoothly and quickly
```

---

## Example — Minimise f(x, y) = (x − 3)² + 2(y + 1)²

```
True minimum is at  x* = 3,  y* = -1

∂f/∂x = 2(x - 3)
∂f/∂y = 4(y + 1)
```

````python
# Exercise 7
import matplotlib.pyplot as plt
import numpy as np


def f(x, y):
    return (x - 3) ** 2 + 2 * (y + 1) ** 2


def gradient_f_ex7(x, y):
    df_dx = 2 * (x - 3)
    df_dy = 4 * (y + 1)
    return np.array([df_dx, df_dy])


def run_descent(alpha, steps=30):
    w = np.array([0.0, 0.0])
    values = [f(*w)]
    for _ in range(steps):
        g = gradient_f_ex7(*w)
        w = w - alpha * g
        values.append(f(*w))
    return values, w


alphas = [0.5, 0.1, 0.01]
colors = ["darkorange", "steelblue", "green"]

fig, ax = plt.subplots(figsize=(8, 5))

for alpha, color in zip(alphas, colors):
    values, final_w = run_descent(alpha)
    ax.plot(
        values,
        color=color,
        lw=2,
        label=f"α={alpha}   =>  end=({final_w[0]:.2f}, {final_w[1]:.2f})",
    )
    print(
        f"α={alpha}    final point: x={final_w[0]:.4f}, y={final_w[1]:.4f},  f={f(*final_w):.6f}"
    )

ax.set_xlabel("Step")
ax.set_ylabel("f(x, y)")
ax.set_title(
    "Effect of Learning Rate on Gradient Descent", fontweight="bold"
)
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\nTrue minimum: x=3.0, y=-1.0  =>  f=0.0")
````
