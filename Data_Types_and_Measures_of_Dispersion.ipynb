## Part B: Central Tendency from Scratch (Lecture 1)

### Task 3: Implement Mean, Median, and Mode

```python
def my_mean(data):
    return sum(data) / len(data)

def my_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2
    if n % 2 == 0:
        return (sorted_data[mid - 1] + sorted_data[mid]) / 2.0
    return float(sorted_data[mid])

def my_mode(data):
    counts = {}
    for item in data:
        counts[item] = counts.get(item, 0) + 1
    max_freq = max(counts.values())
    return [k for k, v in counts.items() if v == max_freq]

print(f"Mean:   {my_mean(scores)}")
print(f"Median: {my_median(scores)}")
print(f"Mode:   {my_mode(scores)}")
```

### Task 4: Outlier Effect on Central Tendency

```python
scores_with_outlier = scores + [500]

print(f"Mean with outlier:   {my_mean(scores_with_outlier):.2f}")
print(f"Median with outlier: {my_median(scores_with_outlier):.2f}")
print(f"Mode with outlier:   {my_mode(scores_with_outlier)}")
print("\n--- Analysis ---")
print("Changed most: Mean (pulled drastically from 78.3 to 116.63)")
print("Changed least: Median / Mode (Median slightly shifted from 80.0 to 82.0)")
print("Why median for salary: Median is robust to extreme high-income outliers that heavily skew the mean upwards.")
```

## Part C: Measures of Dispersion from Scratch (Lecture 2)

### Task 6: Implement Variance & Standard Deviation

```python
def my_variance(data):
    """Sample variance: divide by (n - 1)"""
    mu = my_mean(data)
    return sum((x - mu) ** 2 for x in data) / (len(data) - 1)

def my_std(data):
    """Sample standard deviation"""
    return my_variance(data) ** 0.5

mean_s = my_mean(salaries)
var_s = my_variance(salaries)
std_s = my_std(salaries)

print(f"Mean:          {mean_s:,.0f}")
print(f"Variance:      {var_s:,.0f}")
print(f"Std Dev:       {std_s:,.0f}")
print(f"Typical range: {mean_s - std_s:,.0f} to {mean_s + std_s:,.0f}")
```

### Task 7: Coefficient of Variation (CV)

```python
def my_cv(data):
    """CV = (std / mean) * 100"""
    return (my_std(data) / my_mean(data)) * 100

cv_scores = my_cv(scores)
cv_salaries = my_cv(salaries)

print(f"CV Scores:   {cv_scores:.2f}%")
print(f"CV Salaries: {cv_salaries:.2f}%")
print("\nMore variable dataset: Salaries")
print("Why raw std dev cannot be compared directly: Scores and salaries have completely different units (points vs PKR) and different scales.")
```

### Task 8: Skewness

```python
def my_skewness(data):
    """Pearson's 2nd coefficient: 3 * (mean - median) / std"""
    mu = my_mean(data)
    med = my_median(data)
    std = my_std(data)
    return 3 * (mu - med) / std

sk = my_skewness(salaries)
print(f"Skewness: {sk:.4f}")
print("Interpretation:")
print(" - Direction: Positive")
print(" - Tail direction: Right")
print(" - Relationship: Mean (50,417) is greater than Median (46,500)")
print(" - Best central measure: Median")
print(" - ML Action: Apply log transformation or power transform (e.g., Box-Cox) to reduce right skewness.")
```

### Task 9: Excess Kurtosis

```python
def my_excess_kurtosis(data):
    mu = my_mean(data)
    std = my_std(data)
    n = len(data)
    fourth_moment = sum(((x - mu) / std) ** 4 for x in data) / n
    return fourth_moment - 3

k_scores = my_excess_kurtosis(scores)
k_salaries = my_excess_kurtosis(salaries)

print(f"Scores excess kurtosis:   {k_scores:.4f} (Platykurtic - lighter tails)")
print(f"Salaries excess kurtosis: {k_salaries:.4f} (Leptokurtic - heavy tails)")
print("\nHigh positive excess kurtosis implies high risk of extreme outliers, which can heavily bias models sensitive to extreme values (e.g., Linear Regression).")
```

## Part D: Z-Score, Percentiles & Quartiles (Lecture 2)

### Task 10: Z-Score Outlier Detection

```python
def my_z_scores(data):
    mu = my_mean(data)
    std = my_std(data)
    return [(x - mu) / std for x in data]

z = my_z_scores(salaries)
for val, zi in zip(salaries, z):
    flag = " <-- OUTLIER" if abs(zi) > 2 else ""
    print(f"{val:>8,} z = {zi:+.3f}{flag}")

print("\nVerification with NumPy:")
mu_np = np.mean(salaries)
std_np = np.std(salaries, ddof=1)
z_np = [(x - mu_np) / std_np for x in salaries]
print("NumPy z-scores:", [round(zi, 3) for zi in z_np])
print("Results match? Yes")
```

### Task 11: Percentile Implementation

```python
def my_percentile(data, p):
    sorted_data = sorted(data)
    n = len(sorted_data)
    rank = (p / 100) * (n - 1)
    lower = int(rank)
    upper = lower + 1
    if upper >= n:
        return float(sorted_data[-1])
    frac = rank - lower
    return sorted_data[lower] + frac * (sorted_data[upper] - sorted_data[lower])

for p in [25, 50, 75, 95]:
    val = my_percentile(salaries, p)
    print(f"P{p:2d} = {val:,.0f}")

print("\nDoes P50 match my_median()? Yes (P50 = 46,500, my_median = 46,500.0)")
```

### Task 12: Quartiles, IQR, and Tukey Fences

```python
q1 = my_percentile(salaries, 25)
q3 = my_percentile(salaries, 75)
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr
outliers = [x for x in salaries if x < lower_fence or x > upper_fence]

print(f"Q1:          {q1:>10,.0f}")
print(f"Q3:          {q3:>10,.0f}")
print(f"IQR:         {iqr:>10,.0f}")
print(f"Lower fence: {lower_fence:>10,.0f}")
print(f"Upper fence: {upper_fence:>10,.0f}")
print(f"Outlier(s):  {outliers}")
```

### Task 13: Complete Stats Summary

```python
def stats_summary(data, label="Dataset"):
    n = len(data)
    mu = my_mean(data)
    med = my_median(data)
    mod = my_mode(data)
    sd = my_std(data)
    var = my_variance(data)
    cv = my_cv(data)
    sk = my_skewness(data)
    ek = my_excess_kurtosis(data)
    q1 = my_percentile(data, 25)
    q3 = my_percentile(data, 75)
    iqr_val = q3 - q1
    lo = q1 - 1.5 * iqr_val
    hi = q3 + 1.5 * iqr_val
    out = [x for x in data if x < lo or x > hi]

    print(f"\n{'='*50}")
    print(f" SUMMARY: {label}")
    print(f"{'='*50}")
    print(f" n         : {n}")
    print(f" Min / Max : {min(data)} / {max(data)}")
    print(f" Mean      : {mu:.2f}")
    print(f" Median    : {med:.2f}")
    print(f" Mode      : {mod}")
    print(f" Std Dev   : {sd:.2f}")
    print(f" Variance  : {var:.2f}")
    print(f" CV        : {cv:.2f}%")
    print(f" Skewness  : {sk:.4f}")
    print(f" Exc. Kurt : {ek:.4f}")
    print(f" Q1/Q2/Q3  : {q1:.2f} / {med:.2f} / {q3:.2f}")
    print(f" IQR       : {iqr_val:.2f}")
    print(f" Fences    : [{lo:.2f}, {hi:.2f}]")
    print(f" Outliers  : {out}")

stats_summary(scores, "Exam Scores")
stats_summary(salaries, "Salaries (PKR)")
```

## Reflection Questions

1. **When would you choose median over mean as the central tendency measure for a feature in an ML pipeline? Give a real dataset example.**
   * **Answer:** Median is preferred over mean when a feature contains extreme skewness or strong outliers that would distort the mean. For example, in real estate pricing datasets (e.g., California Housing), a few multi-million-dollar mansions inflate the mean price, making median home value a much more realistic center for data scaling and imputation.

2. **The CV of Dataset A is 4% and Dataset B is 28%. What does this tell you about relative variability? Which dataset is more risky to model without preprocessing?**
   * **Answer:** Dataset B has significantly higher relative variability (28%) compared to Dataset A (4%), meaning its values are much more dispersed relative to its mean scale. Dataset B is more risky to model directly without normalization or scaling because wide variances can destabilize gradient-based estimators.

3. **A feature has skewness = 1.8. What preprocessing transformation would you apply before passing it to a linear regression model, and why?**
   * **Answer:** A positive skewness of 1.8 indicates heavy right skewness. Applying a logarithmic transformation (e.g., `np.log1p`) or a Power Transform (Box-Cox / Yeo-Johnson) would compress high extreme values, normalize the distribution, and satisfy the homoscedasticity and normality assumptions of linear regression.

4. **Your z-score method flags one outlier but Tukey's IQR fence flags none (or vice versa). Which do you trust more for a salary dataset? Why?**
   * **Answer:** Tukey's IQR method is generally more reliable for salary data. Z-scores rely on the mean and standard deviation, which are themselves distorted by the outlier (causing standard deviation to inflate and hide additional extreme values). IQR relies on order-based percentiles ($Q_1$ and $Q_3$), which remain uncorrupted by extreme values.
